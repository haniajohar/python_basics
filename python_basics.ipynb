```python
class teacher:
  def __init__(self,name,age,salary):
    self.name=name
    self.age=age
    self.salary=salary

  def teach(self):
    return f"{self.name} is teaching."

class research:
  def research(self):
    return "Conducting research."

class professor(teacher, research):
  def __init__(self, name, age, salary, department):
    super().__init__(name, age, salary)
    self.department = department

  def guide(self):
    return f"{self.name} is guiding students in the {self.department} department."

professor_obj = professor("Dr. Hania", 45, 100000, "Computer Science")
print(professor_obj.teach())
print(professor_obj.research())
print(professor_obj.guide())
```

```python
def calculate_adjusted_stats(marks):
    adjusted_marks = marks.copy()
    adjusted_marks.extend(marks[:2])
    adjusted_mean = sum(adjusted_marks) / len(adjusted_marks)
    adjusted_min = min(adjusted_marks)
    adjusted_max = max(adjusted_marks)
    sorted_marks = sorted(marks)
    n = len(sorted_marks)
    if n % 2 == 1:
        median = sorted_marks[n // 2]
    else:
        median = (sorted_marks[n // 2 - 1] + sorted_marks[n // 2]) / 2
    from collections import Counter
    freq = Counter(marks)
    max_count = max(freq.values())
    mode = [k for k, v in freq.items() if v == max_count]
    return adjusted_mean, adjusted_min, adjusted_max, median, mode
marks = [12, 15, 18, 10, 16, 14, 17, 11, 15, 13]
adjusted_mean, adj_min, adj_max, median, mode = calculate_adjusted_stats(marks)
print(f"Original marks: {marks}")
print(f"\nWith first two students counted twice:")
print(f"Adjusted mean: {adjusted_mean:.2f}")
print(f"Adjusted minimum: {adj_min}")
print(f"Adjusted maximum: {adj_max}")
print(f"\nUsing original list:")
print(f"Median: {median}")
print(f"Mode: {mode}")
```

```python
num_students = int(input("Enter number of students: "))

averages = []

for i in range(num_students):
    print(f"\nStudent {i+1}")

    name = input("Enter student name: ")

    m1 = float(input("Enter marks in Subject 1: "))
    m2 = float(input("Enter marks in Subject 2: "))
    m3 = float(input("Enter marks in Subject 3: "))

    avg = (m1 + m2 + m3) / 3
    averages.append(avg)

    result = "Pass" if avg >= 50 else "Fail"

    print(f"Average Marks = {avg:.2f}")
    print(f"Result = {result}")

print("\nClass Summary")
print("Highest Average =", max(averages))
print("Lowest Average =", min(averages))
```

```python
import math

def add(a, b):
    return a + b

def diff(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Division by zero"
    return a / b

def square(a):
    return a * a

def cube(a):
    return a ** 3

def factorial(a):
    if a < 0:
        return "Error: Negative factorial not allowed"
    return math.factorial(int(a))

def log10(a):
    if a <= 0:
        return "Error: Log undefined"
    return math.log10(a)

def menu():
    print("\n--- Mathematical Utility ---")
    print("1. Add")
    print("2. Difference")
    print("3. Multiply")
    print("4. Divide")
    print("5. Square")
    print("6. Cube")
    print("7. Factorial")
    print("8. Log10")
    print("9. Exit")

while True:
    menu()
    choice = int(input("Enter choice: "))

    if choice == 9:
        print("Exiting Program...")
        break

    if choice in [1,2,3,4]:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))

    else:
        a = float(input("Enter number: "))

    if choice == 1:
        print(add(a,b))
    elif choice == 2:
        print(diff(a,b))
    elif choice == 3:
        print(multiply(a,b))
    elif choice == 4:
        print(divide(a,b))
    elif choice == 5:
        print(square(a))
    elif choice == 6:
        print(cube(a))
    elif choice == 7:
        print(factorial(a))
    elif choice == 8:
        print(log10(a))
    else:
        print("Invalid Choice")
```

```python
%%writefile my_numbers.py

class number_ops:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def sum(self):
        return self.x + self.y

    def difference(self):
        return self.x - self.y

    def product(self):
        return self.x * self.y

    def quotient(self):
        if self.y == 0:
            return "Division by zero error"
        return self.x / self.y
```

```python
from my_numbers import number_ops
import math

class math_ops(number_ops):

    def absolute(self, value):
        return abs(value)

    def square_root(self, value):
        if value < 0:
            return "Invalid input"
        return math.sqrt(value)

    def power(self):
        return self.x ** self.y
```

```python
x = float(input("Enter first number: "))
y = float(input("Enter second number: "))

obj = math_ops(x, y)

print("\nBasic Operations")
print("Sum =", obj.sum())
print("Difference =", obj.difference())
print("Product =", obj.product())
print("Quotient =", obj.quotient())

print("\nAdvanced Operations")
num = float(input("Enter number for abs and sqrt: "))
print("Absolute =", obj.absolute(num))
print("Square Root =", obj.square_root(num))
print("Power =", obj.power())
```
